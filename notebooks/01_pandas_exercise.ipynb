```python
# Fill missing quantity values with 1 and missing unit_price values with the median. Which orders had missing values?
```

```python
df['quantity']=df['quantity'].fillna(1)
```

```python
df
```

```python
df['unit_price'].isna().sum()
```

```python
df['unit_price']=df['unit_price'].fillna(df['unit_price'].median())
```

```python
df
```

```python
df.isna().sum()
```

```python
# Add a gross_total column equal to quantity * unit_price.
```

```python
df['gross_total']=df['quantity']*df['unit_price']
```

```python
df
```

```python
# Add net_total after discount: gross_total * (1 - discount_pct). Round to 2 decimal places.
```

```python
df['net_total_after_discount']=df['gross_total']*(1-df['discount_pct']).round(2)
```

```python
df
```

```python
# Calculate total net sales by city. Then calculate shipped-only net sales by city.
```

```python
df.groupby('city')['net_total_after_discount'].sum()
```

```python
df[df['status']=='Shipped'].groupby('city')['net_total_after_discount'].sum()
```
